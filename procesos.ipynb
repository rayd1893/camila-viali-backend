```python
import pandas as pd

# Nombre de tu archivo (asegúrate de que esté en el mismo directorio o proporciona la ruta completa)
nombre_archivo = "./Reporte_Mis_depósitos_detalle_18092025181529.xlsx"

# Leer el archivo de Excel
# header=6 indica que la fila con índice 6 (la Fila 7 de Excel) se usará como encabezado
try:
    df_excel = pd.read_excel(
        nombre_archivo,
        header=7
    )
    
    print("DataFrame creado exitosamente a partir de la Fila 7 (índice 6).")
    print("\nPrimeras 5 filas del DataFrame:")
    print(df_excel.head())
    print("\nNombres de las columnas (tomados de la Fila 7):")
    print(df_excel.columns.tolist())

except FileNotFoundError:
    print(f"Error: No se encontró el archivo '{nombre_archivo}'. Verifica la ruta.")
except Exception as e:
    print(f"Ocurrió un error al leer el archivo: {e}")
```

```python
from dotenv import load_dotenv
import pandas as pd
from sqlalchemy import create_engine
import os


load_dotenv()

# --- PARÁMETROS DE LA CONEXIÓN MYSQL ---
DB_USER = os.getenv('MYSQL_USER')        # Tu nombre de usuario de MySQL
DB_PASSWORD = os.getenv('MYSQL_PASSWORD')         # Tu contraseña de MySQL
DB_HOST = os.getenv('MYSQL_HOST')                 # La dirección del servidor (o IP)
DB_PORT = "3306"                    # El puerto estándar de MySQL
DB_NAME = os.getenv('MYSQL_DB') # El nombre de la base de datos de destino

# La cadena de conexión sigue el formato:
# dialect+driver://user:password@host:port/database
engine = create_engine('mysql+mysqldb://{}:{}@{}/{}'.format(DB_USER, DB_PASSWORD, DB_HOST, DB_NAME))

# --- Suponiendo que 'df_excel' es tu DataFrame de Pandas ---
# Si tu DataFrame se llama diferente, ajústalo aquí.
df_a_cargar = df_excel.copy()
```

```python
# --- PARÁMETROS DE CARGA ---
NOMBRE_TABLA_DESTINO = "base_niubiz"

try:
    df_a_cargar.to_sql(
        name=NOMBRE_TABLA_DESTINO,
        con=engine,
        if_exists='replace', # Opciones: 'fail', 'replace', 'append'
        index=False
    )
    
    print(f"✅ Éxito: El DataFrame se cargó en la tabla '{NOMBRE_TABLA_DESTINO}' de MySQL.")

except Exception as e:
    print(f"❌ Error al cargar el DataFrame a MySQL: {e}")

finally:
    # Cerrar la conexión
    engine.dispose()
```

```python
from datetime import datetime
import pytz
def convert_timestamp_peru(timestamp):
  my_timestamp = datetime.fromtimestamp(timestamp)
  old_timezone = pytz.timezone("America/Lima")
  new_timezone = pytz.timezone("UTC")
  localized_timestamp = old_timezone.localize(my_timestamp)
  new_timezone_timestamp = localized_timestamp.astimezone(new_timezone)
  return new_timezone_timestamp

convert_timestamp_peru(1756684800)
```
